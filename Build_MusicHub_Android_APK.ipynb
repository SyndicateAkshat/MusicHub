# 📱 MusicHub Android APK Generator (Google Colab)
Run this notebook in Google Colab to automatically build `MusicHub-1.0.0-arm64-v8a-debug.apk`!

```python
# 1. Install Buildozer and Linux compilation tools
!apt-get update
!apt-get install -y build-essential git python3-pip ccache libffi-dev libssl-dev libxml2-dev libxslt1-dev zlib1g-dev openjdk-17-jdk unzip zip
!pip install --upgrade buildozer cython kivy
```

```python
# 2. Build the Android APK
!buildozer -v android debug
```

```python
# 3. Download the generated APK file to your phone/computer
from google.colab import files
import glob

apks = glob.glob('bin/*.apk')
if apks:
    print(f'Downloading APK: {apks[0]}')
    files.download(apks[0])
else:
    print('No APK found in bin/ directory.')
```
